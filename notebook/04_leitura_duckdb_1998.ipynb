### 9.obter as variáveis de 1998

```python
def obter_variaveis_ano(
    catalogo: pd.DataFrame,
    ano: int,
) -> list[str]:
    """
    Retorna as variáveis selecionadas para determinado ano,
    excluindo células vazias e títulos de seção.
    """
    coluna_ano = str(ano)

    if coluna_ano not in catalogo.columns:
        raise KeyError(
            f"O ano {ano} não existe no catálogo."
        )

    valores = (
        catalogo[coluna_ano]
        .dropna()
        .str.strip()
        .tolist()
    )

    variaveis = []

    for valor in valores:
        if not valor:
            continue

        if valor.upper().startswith("VARIÁVEIS "):
            continue

        if valor.lower() == "nome da variável":
            continue

        variaveis.append(valor)

    return variaveis
```

```python
variaveis_1998 = obter_variaveis_ano(
    catalogo=catalogo,
    ano=ano_teste,
)

print(
    f"Variáveis selecionadas para {ano_teste}: "
    f"{len(variaveis_1998)}"
)

for numero, variavel in enumerate(
    variaveis_1998,
    start=1,
):
    print(f"{numero:02d} - {variavel}")
```

### 18. Definir Grupo de Variaveis

```python

variaveis_numericas = [
    "NU_ANO",
    "TP_FAIXA_ETARIA",
    "TP_PRESENCA",
    "VL_PERC_COMP1",
    "VL_PERC_COMP2",
    "VL_PERC_COMP3",
    "VL_PERC_COMP4",
    "VL_PERC_COMP5",
    "NU_NOTA_OBJETIVA",
    "NU_NOTA_COMP1",
    "NU_NOTA_COMP2",
    "NU_NOTA_COMP3",
    "NU_NOTA_COMP4",
    "NU_NOTA_COMP5",
    "NU_NOTA_REDACAO",
]

variaveis_texto = [
    variavel
    for variavel in variaveis_1998
    if variavel not in variaveis_numericas
]

print(f"Variáveis numéricas: {len(variaveis_numericas)}")
print(f"Variáveis de texto: {len(variaveis_texto)}")
```

### 19.Buscar por  valores numéricos inválidos

```python
expressoes_invalidos = []

for variavel in variaveis_numericas:
    coluna = citar_identificador(variavel)

    expressao = f"""
        SUM(
            CASE
                WHEN NULLIF(TRIM({coluna}), '') IS NOT NULL
                 AND TRY_CAST({coluna} AS DOUBLE) IS NULL
                THEN 1
                ELSE 0
            END
        ) AS {coluna}
    """

    expressoes_invalidos.append(expressao)

sql_invalidos = ",\n".join(
    expressoes_invalidos
)

consulta_invalidos = f"""
SELECT
    {sql_invalidos}

FROM enem_1998_raw

WHERE SG_UF_RESIDENCIA = 'PA'
"""

invalidos_numericos = conexao.execute(
    consulta_invalidos
).df()

invalidos_numericos
```

### 22-Definir Tipos

```python
variaveis_inteiras = {
    "NU_ANO",
    "TP_FAIXA_ETARIA",
    "TP_PRESENCA",
}

variaveis_decimais = {
    "VL_PERC_COMP1",
    "VL_PERC_COMP2",
    "VL_PERC_COMP3",
    "VL_PERC_COMP4",
    "VL_PERC_COMP5",
    "NU_NOTA_OBJETIVA",
    "NU_NOTA_COMP1",
    "NU_NOTA_COMP2",
    "NU_NOTA_COMP3",
    "NU_NOTA_COMP4",
    "NU_NOTA_COMP5",
    "NU_NOTA_REDACAO",
}
```

### 23.Construir as conversões

```python
expressoes_transformacao = []

for variavel in variaveis_1998:
    coluna = citar_identificador(variavel)

    if variavel in variaveis_inteiras:
        expressao = (
            f"TRY_CAST(NULLIF(TRIM({coluna}), '') "
            f"AS SMALLINT) AS {coluna}"
        )

    elif variavel in variaveis_decimais:
        expressao = (
            f"TRY_CAST(NULLIF(TRIM({coluna}), '') "
            f"AS DOUBLE) AS {coluna}"
        )

    else:
        expressao = (
            f"NULLIF(TRIM({coluna}), '') "
            f"AS {coluna}"
        )

    expressoes_transformacao.append(expressao)

transformacoes_sql = ",\n    ".join(
    expressoes_transformacao
)

print(transformacoes_sql)
```

### 24. Criar a view tipada do Pará

```python
conexao.execute(
    f"""
    CREATE OR REPLACE TEMP VIEW enem_1998_pa_tipado AS

    SELECT
        {transformacoes_sql}

    FROM enem_1998_raw

    WHERE SG_UF_RESIDENCIA = 'PA'
    """
)

print("View tipada do Pará criada.")
```

### 28.contar valores ausentes

```python
expressoes_nulos = []

for variavel in variaveis_1998:
    coluna = citar_identificador(variavel)

    expressoes_nulos.append(
        f"""
        COUNT(*) FILTER (
            WHERE {coluna} IS NULL
        ) AS {coluna}
        """
    )

consulta_nulos = f"""
SELECT
    {", ".join(expressoes_nulos)}

FROM enem_1998_pa_tipado
"""

nulos = conexao.execute(
    consulta_nulos
).df()

nulos_vertical = (
    nulos
    .T
    .reset_index()
)

nulos_vertical.columns = [
    "variavel",
    "quantidade_nulos",
]

nulos_vertical[
    nulos_vertical["quantidade_nulos"] > 0
]
```

### 34.definir o destino Silver

```python
SILVER_DIR = (
    PROJECT_ROOT
    / "data"
    / "silver"
    / "microdados_selecionados"
)

diretorio_silver = (
    SILVER_DIR
    / f"ano={ano_teste}"
    / "uf=PA"
)

arquivo_parquet = (
    diretorio_silver
    / f"enem_PA_{ano_teste}.parquet"
)

print(f"Diretório: {diretorio_silver}")
print(f"Parquet  : {arquivo_parquet}")
```

### 35.preparar o diretório

```python
diretorio_silver.mkdir(
    parents=True,
    exist_ok=True,
)

if arquivo_parquet.exists():
    raise FileExistsError(
        f"O arquivo já existe: {arquivo_parquet}"
    )

print("Destino disponível para gravação.")
```

### 36.gravar o Parquet

```python
caminho_parquet_sql = str(
    arquivo_parquet
).replace("'", "''")

conexao.execute(
    f"""
    COPY (
        SELECT *
        FROM enem_1998_pa_tipado
    )
    TO '{caminho_parquet_sql}'
    (
        FORMAT PARQUET,
        COMPRESSION ZSTD
    )
    """
)

print(f"Parquet gravado: {arquivo_parquet}")
```

### 37 — verificar o arquivo

```python
if not arquivo_parquet.exists():
    raise FileNotFoundError(
        "O Parquet não foi criado."
    )

tamanho_parquet_mb = (
    arquivo_parquet.stat().st_size
    / 1024**2
)

print(f"Tamanho: {tamanho_parquet_mb:.4f} MB")
```

### 38 — validar estrutura e quantidade

```python
validacao_parquet = conexao.execute(
    f"""
    SELECT
        COUNT(*) AS total_linhas,
        COUNT(*) - COUNT(DISTINCT NU_INSCRICAO)
            AS duplicacoes,
        COUNT(*) FILTER (
            WHERE SG_UF_RESIDENCIA <> 'PA'
        ) AS outras_ufs,
        MIN(NU_ANO) AS menor_ano,
        MAX(NU_ANO) AS maior_ano

    FROM read_parquet(
        '{caminho_parquet_sql}',
        hive_partitioning = false
    )
    """
).df()

validacao_parquet.T
```

### 39 — comparar origem e Parquet

```python
comparacao = conexao.execute(
    f"""
    SELECT COUNT(*) AS quantidade_diferencas

    FROM (
        (
            SELECT *
            FROM enem_1998_pa_tipado

            EXCEPT ALL

            SELECT *
            FROM read_parquet(
                '{caminho_parquet_sql}',
                hive_partitioning = false
            )
        )

        UNION ALL

        (
            SELECT *
            FROM read_parquet(
                '{caminho_parquet_sql}',
                hive_partitioning = false
            )

            EXCEPT ALL

            SELECT *
            FROM enem_1998_pa_tipado
        )
    )
    """
).df()

comparacao.T
```
